```python
# Q33. Find the Sum of Even Numbers from 1 to N

n = int(input("Enter a number: "))

sum = 0

for i in range(2, n + 1, 2):
    sum += i

print("Sum =", sum)

```

```python
# Q34. Find the Sum of Odd Numbers from 1 to N

n = int(input("Enter a number: "))

sum = 0

for i in range(1, n + 1, 2):
    sum += i

print("Sum =", sum)

```

```python
# Q35. Print the Fibonacci Series

n = int(input("Enter the number of terms: "))

a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    c = a + b
    a = b
    b = c
```

```python
# Q36. Find the Sum of Digits

num = int(input("Enter a number: "))

sum = 0

while num > 0:
    digit = num % 10
    sum += digit
    num //= 10

print("Sum of Digits =", sum)
```

```python
# Q37. Find the Product of Digits

num = int(input("Enter a number: "))

product = 1

while num > 0:
    digit = num % 10
    product *= digit
    num //= 10

print("Product of Digits =", product)
```

```python
# Q38. Check Whether a Number is Perfect
num = int(input("Enter a number: "))

sum = 0

for i in range(1, num):
    if num % i == 0:
        sum += i

if sum == num:
    print("Perfect Number")
else:
    print("Not a Perfect Number")
```

```python
# Q39. Find All Factors of a Number

num = int(input("Enter a number: "))

print("Factors are:")

for i in range(1, num + 1):
    if num % i == 0:
        print(i)
```

```python
# Q40. Find the Greatest Common Divisor (GCD)

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

while b != 0:
    a, b = b, a % b

print("GCD =", a)
```
