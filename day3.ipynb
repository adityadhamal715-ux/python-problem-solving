```python
# Q23. Print the Multiplication Table of a Number

num = int(input("Enter a number: "))

for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

```python
# Q24. Find the Sum of First N Natural Numbers

n = int(input("Enter a number: "))

sum = 0

for i in range(1, n + 1):
    sum += i

print("Sum =", sum)
```

```python
# Q25. Find the Factorial of a Number

n = int(input("Enter a number: "))

fact = 1

for i in range(1, n + 1):
    fact *= i

print("Factorial =", fact)

```

```python
# Q26. Count the Number of Digits

num = int(input("Enter a number: "))

count = len(str(abs(num)))

print("Total Digits =", count)
```

```python
# Q27. Reverse a Number

num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10

print("Reverse =", reverse)
```

```python
# Q28. Check Whether a Number is Palindrome

num = int(input("Enter a number: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10

if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Q29. Check Whether a Number is Armstrong

num = int(input("Enter a number: "))

temp = num
digits = len(str(num))
total = 0

while temp > 0:
    digit = temp % 10
    total += digit ** digits
    temp //= 10

if total == num:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")

```

```python
# Q30. Check Whether a Number is Prime

num = int(input("Enter a number: "))

if num > 1:
    prime = True

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            prime = False
            break

    if prime:
        print("Prime Number")
    else:
        print("Not a Prime Number")
else:
    print("Not a Prime Number")
```
